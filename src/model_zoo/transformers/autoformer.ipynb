# **Autoformer**

Это ещё один «специализированный трансформер» для временных рядов, близкий по духу к FEDformer, но с другим ключевым механизмом: auto-correlation вместо классического self-attention + явная декомпозиция тренд+сезонность на каждом слое.

R(τ) = ∑ₜ xₜ · xₜ₋τ

τ — лаг; R(τ) — автокорреляция.
