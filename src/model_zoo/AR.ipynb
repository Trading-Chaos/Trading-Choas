# Авторегрессия (AR) 

— это метод прогнозирования, который основывается на идее, что текущее значение временного ряда можно выразить как линейную комбинацию его предыдущих значений. Эта модель активно используется для анализа и прогнозирования временных рядов, например, в экономике, финансах и других областях, где данные демонстрируют определенную последовательность.

## Основные концепции авторегрессии:
	1.	Стационарность: Важнейшее требование для модели AR. Это означает, что среднее и дисперсия временного ряда остаются неизменными на протяжении времени. Если данные не являются стационарными, их необходимо привести к стационарному виду, например, путем дифференцирования.
	2.	Порядок модели AR(p): В модели AR(p), где p — это порядок модели, текущее значение зависит от p предыдущих значений временного ряда. Например, AR(1) использует только одно предыдущее значение, AR(2) — два, и так далее.
	3.	Формула AR(p):
$$
X_t = \phi_1 X_{t-1} + \phi_2 X_{t-2} + \dots + \phi_p X_{t-p} + \epsilon_t
$$
где:
	•	$$ X_t $$ — текущее значение,
	•	$$ \phi_1, \phi_2, \dots, \phi_p $$ — коэффициенты модели,
	•	$$ \epsilon_t $$ — ошибка модели (шум).
	4.	AR(1): Простейшая форма авторегрессии, где текущее значение зависит только от одного предыдущего значения:
$$
X_t = \phi_1 X_{t-1} + \epsilon_t
$$ 
Здесь $ \phi_1 $ — коэффициент, определяющий, насколько сильно текущее значение зависит от прошлого.
	5.	Оценка параметров: Для определения коэффициентов модели \phi используются различные методы, такие как метод наименьших квадратов (МНК) или метод максимального правдоподобия (ММП).
	6.	Тестирование на стационарность: Перед использованием модели AR необходимо проверить стационарность ряда с помощью ADF-теста (Augmented Dickey-Fuller). Если p-value теста меньше 0.05, то ряд стационарен и можно применять AR-модель. Если ряд нестационарен, его нужно дифференцировать.
	7.	Диагностика модели: После построения модели важно провести диагностику остатков, чтобы убедиться, что ошибка случайна и не содержит значимых паттернов.
