```python
ARTIFACT_DIR = (
    PROJECT_ROOT
    / "src"
    / "MLbridg"
    / "artifacts"
    / "xgb_ttp"
)
```

```python
import joblib
import json
from pathlib import Path
from xgboost import XGBClassifier

ARTIFACT_DIR.mkdir(parents=True, exist_ok=True)

def train_xgb_ttp(
    df,
    train_size,
    test_size,
    step,
):

    splitter = prep(
        df=df,
        target_fn=ttp_target,
        target_name="ttp",
        target_col="TTP_class",
        horizons=[12, 24, 48],
        train_size=train_size,
        test_size=test_size,
        step=step,
        target_kwargs={"n_classes": 3},
        scale_cols=[
            "Open", "High", "Low", "Close",
            "Alligator_Jaw", "Alligator_Teeth", "Alligator_Lips",
            "AO",
            "AddOn_Anchor_Level", "AddOn_Size_Pct"
        ]
    )

    best_model = None
    best_scaler = None
    best_score = -1.0

    for fold_id, (X_train, X_test, y_train, y_test, scaler) in enumerate(splitter):

        model = XGBClassifier(
            objective="multi:softprob",
            num_class=3,
            n_estimators=300,
            max_depth=5,
            learning_rate=0.05,
            subsample=0.8,
            colsample_bytree=0.8,
            tree_method="hist",
            random_state=42
        )

        model.fit(X_train, y_train)
        y_pred = model.predict(X_test)

        metrics = merged_metrics(y_test, y_pred)

        append_results({
            "task_type": "classification",
            "model_name": "XGBoost",
            "model_family": "xgb",
            "fold": fold_id,
            "model_params": {
                "n_estimators": 300,
                "max_depth": 5,
                "learning_rate": 0.05,
                "subsample": 0.8,
                "colsample_bytree": 0.8
            },
            "target_name": "ttp",
            "target_variant": "3class",
            "horizons": "12_24_48",
            **metrics
        })

        # 👉 критерий отбора модели (можешь заменить)
        score = metrics.get("balanced_accuracy", 0.0)

        if score > best_score:
            best_score = score
            best_model = model
            best_scaler = scaler

    # ===== SAVE ARTIFACTS =====
    joblib.dump(best_model, ARTIFACT_DIR / "model.pkl")
    joblib.dump(best_scaler, ARTIFACT_DIR / "scaler.pkl")
    FEATURES = list(X_train.columns)
    joblib.dump(FEATURES, ARTIFACT_DIR / "features.pkl")
    print("CWD:", Path.cwd())
    print("Saving to:", ARTIFACT_DIR.resolve())
train_xgb_ttp(
    df=df,
    train_size=1000,
    test_size=200,
    step=100
)
```
